Q1 Write a Python script to load the Superstore dataset from a CSV file into MongoDB

```python
import pandas as pd
from pymongo import MongoClient
```

```python
# MongoDB connection
client = MongoClient("mongodb://localhost:27017/")
db = client["SuperstoreDB"]
orders_collection = db["Orders"]

# Load CSV data into MongoDB
csv_file = "Superstore.csv"  # Update with the actual path

# Read CSV into Pandas DataFrame
df = pd.read_csv(csv_file,encoding = 'latin1')

```

```python
#1 Convert DataFrame to dictionary format and insert into MongoDB
orders_collection.insert_many(df.to_dict(orient="records"))
```

Q2 Retrieve and print all documents from the Orders collection

```python
# 2. Retrieve and print all documents from the Orders collection
all_orders = orders_collection.find()
for order in all_orders:
    print(f"{order}\n")
```

```python
Q3 Count and display the total number of documents in the Orders collection
```

```python
# 3. Count and display the total number of documents in the Orders collection
total_orders = orders_collection.count_documents({})
print(f"Total number of orders: {total_orders}")
```

```python
Q4 Write a query to fetch all orders from the "West" region
```

```python
# 4. Fetch all orders from the "West" region
west_orders = orders_collection.find({"Region": "West"})
for order in west_orders:
    print(f"{order}\n")
```

Q5 Write a query to find orders where Sales is greater than 500

```python
# 5. Find orders where Sales is greater than 500
high_sales_orders = orders_collection.find({"Sales": {"$gt": 500}})
for order in high_sales_orders:
    print(f"{order}\n")
```

Q6 Fetch the top 3 orders with the highest Profit

```python
# 6. Fetch the top 3 orders with the highest Profit
top_profit_orders = orders_collection.find().sort("Profit", -1).limit(3)
for order in top_profit_orders:
    print(f"{order}\n")
```

Q7 Update all orders with Ship Mode as "First Class" to "Premium Class.

```python
# 7. Update all orders with Ship Mode as "First Class" to "Premium Class"
orders_collection.update_many({"Ship Mode": "First Class"}, {"$set": {"Ship Mode": "Premium Class"}})
```

Q8. Delete all orders where Sales is less than 50

```python
#8. Delete all orders where Sales is less than 50
orders_collection.delete_many({"Sales": {"$lt": 50}})
```

Q9 Use aggregation to group orders by Region and calculate total sales per region

```python
# 9. Group orders by Region and calculate total sales per region
pipeline = [
    {"$group": {"_id": "$Region", "TotalSales": {"$sum": "$Sales"}}}
]
region_sales = orders_collection.aggregate(pipeline)
for region in region_sales:
    print(f"{region}\n")
```

Q10 Fetch all distinct values for Ship Mode from the collection

```python
#10 . Fetch all distinct values for Ship Mode
distinct_ship_modes = orders_collection.distinct("Ship Mode")
print("Distinct Ship Modes:", distinct_ship_modes)
```

Q11 Count the number of orders for each category.

```python
# 11. Count the number of orders for each category
category_count = orders_collection.aggregate([
    {"$group": {"_id": "$Category", "OrderCount": {"$sum": 1}}}
])
print(f"\n Number of orders for each catogory \n")
for category in category_count:
    print(f"{category}")
```
